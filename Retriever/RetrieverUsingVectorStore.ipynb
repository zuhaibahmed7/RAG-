```python
from langchain_core.documents import Document
document = [
    Document(page_content=(
        "An embedding is a list of numbers that represents the meaning of a piece of text. "
        "Texts with similar meanings have embeddings that are close together, which makes "
        "semantic search possible."
    )),
    Document(page_content=(
        "A vector store is a database that stores embeddings together with the original text "
        "and metadata. Examples include Chroma, Pinecone and FAISS. It finds the stored "
        "vectors closest to a query vector."
    )),
    Document(page_content=(
        "A retriever takes a question and returns the most relevant documents. "
        "In LangChain you can turn any vector store into a retriever with the "
        "as_retriever method."
    )),
    Document(page_content=(
        "Chunking splits long documents into smaller pieces before embedding them. "
        "A chunk_overlap repeats some text between neighbouring chunks, so ideas cut "
        "at a boundary are not lost."
    )),
    Document(page_content=(
        "Hallucination happens when a language model gives a confident answer that is not true. "
        "RAG reduces it by giving the model retrieved documents to base its answer on."
    )),
]
```

```python
!pip install -q langchain-huggingface sentence-transformers langchain-chroma
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name= "sentence-transformers/all-MiniLM-L6-v2")
```

```python
vector_store = Chroma.from_documents(
    documents = document,
    embedding = embeddings,
    collection_name = "my_collection"
)
```

```python
retriever = vector_store.as_retriever(search_kwargs = { 'k' : 2 })
```

```python
for d in retriever.invoke("How do I stop an LLM from making things up?"):
    print("-", d.page_content)
```

```python
# this is the simple similar search but using vector_store.invoke we can use this in the chains and can used in advaced techniques

results2 = vector_store.similarity_search(query = "How do I stop an LLM from making things up?", k = 2)
print(results2)

for i, doc in enumerate(results2):
  print(f"Rsult {i+1}: ")
  print(doc.page_content)
```
